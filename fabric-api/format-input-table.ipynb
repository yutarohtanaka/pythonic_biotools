```python
import pandas as pd
```

```python
df = pd.read_excel('DFMIT Consolidated Run List.xlsx', header = 1)
terra_input_table = df[['Run ID']].rename(columns = {'Run ID': 'entity:fabric_run_id'})
terra_input_table['entity:fabric_run_id'] = terra_input_table['entity:fabric_run_id'].astype(str)
terra_input_table['fabric_run'] = terra_input_table['entity:fabric_run_id'].apply(lambda x: '"{x}"'.format(x=x))
terra_input_table.to_csv('terra_input_table.tsv', index = False, sep = '\t')
terra_input_table.head(1)
```
